## Import Required Libraries

```python
import pandas as pd
```

## Load the Dataset

```python
# Load dataset
# Replace 'Sales_dataset.csv' with your actual file name

df = pd.read_csv('Sales_dataset.csv')

# Display first 5 rows
print(df.head())
```

---

# Q1 Dataset Exploration

## Code

```python
# Display first 10 rows
print("First 10 Rows:")
print(df.head(10))

# Dataset structure
print("\nDataset Info:")
print(df.info())

# Number of rows and columns
print("\nShape of Dataset:")
print(df.shape)

# Data types of columns
print("\nData Types:")
print(df.dtypes)
```

## Expected Insights

* `head(10)` helps understand the dataset structure.
* `info()` displays column names, non-null values, and data types.
* `shape` gives total rows and columns.
* `dtypes` identifies numerical and categorical columns.

---

# Q2 Sales Filtering Analysis

## Code

```python
# Orders where Sales > 3500
high_sales = df[df['Sales'] > 3500]
print("Orders where Sales > 3500")
print(high_sales)

# Orders where Region = 'East' and Sales > 2500
filtered_orders = df[(df['Region'] == 'East') & (df['Sales'] > 2500)]
print("\nEast Region Orders with Sales > 2500")
print(filtered_orders)
```

## Insights

* High-value sales orders contribute significantly to total revenue.
* The East region may contain premium customers or strong-performing products.
* Filtering helps identify important business patterns.

---

# Q3 Customer Purchase Analysis

## Code

```python
# Top 5 customers based on total sales
customer_sales = df.groupby('Customer')['Sales'].sum()

# Sort values in descending order
customer_sales = customer_sales.sort_values(ascending=False)

# Top 5 customers
print("Top 5 Customers by Sales")
print(customer_sales.head(5))
```

## Insights

* Top customers are key revenue contributors.
* Businesses should focus on customer retention strategies.
* Loyalty programs can help maintain these customers.

---

# Q4 Regional Sales Performance

## Code

```python
# Total sales by region
regional_sales = df.groupby('Region')['Sales'].sum()

print("Total Sales by Region")
print(regional_sales)

# Region with highest revenue
highest_region = regional_sales.idxmax()
max_sales = regional_sales.max()

print("\nHighest Revenue Region:")
print(highest_region)
print("Revenue:", max_sales)
```

## Insights

* The region with maximum sales is the strongest market.
* Underperforming regions may need marketing improvements.
* Regional analysis helps in resource allocation.

---

# Q5 Product Performance Analysis

## Code

```python
# Total quantity sold for each product
product_quantity = df.groupby('Product')['Quantity'].sum()

print("Total Quantity Sold by Product")
print(product_quantity)

# Most popular product
most_popular = product_quantity.idxmax()

print("\nMost Popular Product:")
print(most_popular)
```

## Insights

* Products with higher quantities indicate stronger customer demand.
* Popular products should be maintained in inventory.
* Low-performing products may require promotions.

---

# Q6 Multi-Level Grouping

## Code

```python
# Group by Region and Category
region_category_sales = df.groupby(['Region', 'Category'])['Sales'].sum()

print("Sales by Region and Category")
print(region_category_sales)
```

## Insights

* Some categories perform better in specific regions.
* Businesses can optimize regional product strategies.
* Helps identify profitable market segments.

---

# Q7 Pivot Table Creation

## Code

```python
# Create pivot table
pivot_table = pd.pivot_table(
    df,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum'
)

print("Pivot Table")
print(pivot_table)

# Region selling most laptops
if 'Laptop' in pivot_table.columns:
    laptop_region = pivot_table['Laptop'].idxmax()
    print("\nRegion Selling Most Laptops:")
    print(laptop_region)
```

## Insights

* Pivot tables provide summarized business reports.
* Helps compare product performance across regions.
* Useful for decision-making and visualization.

---

# Q8 Profitability Analysis

## Code

```python
# Average profit by category
avg_profit = df.groupby('Category')['Profit'].mean()

print("Average Profit by Category")
print(avg_profit)

# Most profitable category
most_profitable = avg_profit.idxmax()

print("\nMost Profitable Category:")
print(most_profitable)
```

## Insights

* The most profitable category generates the highest average returns.
* Companies should invest more in profitable categories.
* Loss-making categories may need pricing adjustments.

---

# Q9 Discount Impact Study

## Code

```python
# Average profit by discount level
profit_discount = df.groupby('Discount')['Profit'].mean()

print("Average Profit by Discount")
print(profit_discount)
```

## Additional Analysis

```python
# Correlation between Discount and Profit
correlation = df['Discount'].corr(df['Profit'])

print("\nCorrelation between Discount and Profit:")
print(correlation)
```

## Insights

* Higher discounts often reduce profit margins.
* A negative correlation suggests discounts hurt profitability.
* Businesses should balance discounts carefully.

---

# Q10 End-to-End Data Wrangling Task

## Code

```python
# Step 1: Filter dataset where Sales > 3000
filtered_df = df[df['Sales'] > 3000]

# Step 2: Group by Region and Product
sales_summary = filtered_df.groupby(['Region', 'Product'])['Sales'].sum()

# Step 3: Convert into pivot table
pivot_result = sales_summary.unstack()

print("Final Pivot Table")
print(pivot_result)
```

## Business Insights

* High-value sales are concentrated in specific regions and products.
* Certain products dominate premium sales categories.
* Regional demand differences can guide inventory planning.
* Pivot tables simplify strategic business analysis.
